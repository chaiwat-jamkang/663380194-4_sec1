# ประโยชน์ของการเขียนโปรแกรมแบบโมดูลาร์และการสำรวจไลบรารีมาตรฐาน

# 1. ประโยชน์ของการแบ่งโค้ดออกเป็นโมดูล

การแบ่งโค้ดออกเป็นฟังก์ชันและโมดูลมีประโยชน์หลายด้าน ได้แก่

- ทำให้โค้ดอ่านง่าย เพราะแต่ละฟังก์ชันมีหน้าที่เฉพาะ

- ลดการเขียนโค้ดซ้ำ สามารถเรียกใช้ฟังก์ชันเดิมได้หลายครั้ง

- แก้ไขและดีบักได้ง่าย เพราะสามารถตรวจสอบทีละฟังก์ชัน

- เหมาะกับการทำงานเป็นทีม สมาชิกแต่ละคนสามารถรับผิดชอบคนละโมดูลได้

- นำกลับมาใช้ใหม่ได้ ในโปรแกรมอื่นโดยไม่ต้องเขียนใหม่

# 2. การสำรวจไลบรารีมาตรฐานของ Python

# 1) datetime

ใช้สำหรับจัดการวันที่และเวลา

ตัวอย่างการใช้งาน

- แสดงวันที่และเวลาปัจจุบัน

- คำนวณความแตกต่างของวันและเวลา

- จัดรูปแบบวันที่

# 2) os

ใช้สำหรับจัดการไฟล์และโฟลเดอร์

ตัวอย่างการใช้งาน

- ตรวจสอบไฟล์และโฟลเดอร์

- สร้างหรือลบโฟลเดอร์

- ดูตำแหน่งโฟลเดอร์ปัจจุบัน

# 3) collections

ใช้สำหรับจัดการโครงสร้างข้อมูลที่มีประสิทธิภาพมากขึ้น

ตัวอย่างการใช้งาน

- นับจำนวนข้อมูลด้วย Counter

- จัดกลุ่มข้อมูลด้วย defaultdict

- ใช้ deque สำหรับเพิ่มและลบข้อมูลได้รวดเร็ว

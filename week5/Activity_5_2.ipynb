# Activity 5.2: Word Frequency Counter (using Dictionary)

```python
# Activity 5.2 : Word Frequency Counter

# รับข้อความจากผู้ใช้
sentence = input("Enter a sentence: ")

# แปลงเป็นตัวพิมพ์เล็ก
sentence = sentence.lower()

# แยกข้อความออกเป็นคำ
words = sentence.split()

# สร้าง Dictionary ว่าง
word_counts = {}

# วนลูปนับจำนวนคำ
for word in words:
    # ลบเครื่องหมายวรรคตอน
    word = word.strip('.,!?"')

    # ตรวจสอบว่าคำมีอยู่ใน Dictionary หรือไม่
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

# แสดงผลลัพธ์
print("\nWord Frequency:")
print(word_counts)
```

**ตัวอย่างการทำงาน**

Input

```python
Enter a sentence: Python is fun. Python is easy, and Python is powerful!
```

Output

```python
Word Frequency:
{'python': 3, 'is': 3, 'fun': 1, 'easy': 1, 'and': 1, 'powerful': 1}
```

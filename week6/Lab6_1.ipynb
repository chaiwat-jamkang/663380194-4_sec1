```python
# ==========================
# ห้องปฏิบัติการ 6.1: ฟังก์ชันเครื่องคิดเลขอย่างง่าย
# ==========================

def add(a, b):
    """คืนค่าผลบวกของตัวเลขสองจำนวน"""
    return a + b


def subtract(a, b):
    """คืนค่าผลลบของตัวเลขสองจำนวน"""
    return a - b


def multiply(a, b):
    """คืนค่าผลคูณของตัวเลขสองจำนวน"""
    return a * b


def divide(a, b):
    """คืนค่าผลหารของตัวเลขสองจำนวน พร้อมตรวจสอบการหารด้วยศูนย์"""
    if b == 0:
        return "ข้อผิดพลาด: ไม่สามารถหารด้วยศูนย์ได้"
    return a / b


def power(base, exponent=2):
    """คืนค่าผลลัพธ์ของการยกกำลัง โดยค่าเริ่มต้นยกกำลัง 2"""
    return base ** exponent


while True:
    print("\n===== เครื่องคิดเลขอย่างง่าย =====")
    print("1. บวก")
    print("2. ลบ")
    print("3. คูณ")
    print("4. หาร")
    print("5. ยกกำลัง")
    print("0. ออกจากโปรแกรม")

    choice = input("กรุณาเลือกเมนู: ")

    if choice == "0":
        print("ขอบคุณที่ใช้งานโปรแกรม")
        break

    elif choice in ["1", "2", "3", "4"]:
        num1 = float(input("กรุณากรอกตัวเลขตัวที่ 1: "))
        num2 = float(input("กรุณากรอกตัวเลขตัวที่ 2: "))

        if choice == "1":
            result = add(num1, num2)
        elif choice == "2":
            result = subtract(num1, num2)
        elif choice == "3":
            result = multiply(num1, num2)
        elif choice == "4":
            result = divide(num1, num2)

        print("ผลลัพธ์ =", result)

    elif choice == "5":
        base = float(input("กรุณากรอกเลขฐาน: "))

        exponent = input("กรุณากรอกเลขชี้กำลัง (กด Enter หากต้องการใช้ค่าเริ่มต้น = 2): ")

        if exponent == "":
            result = power(base)
        else:
            result = power(base, float(exponent))

        print("ผลลัพธ์ =", result)

    else:
        print("กรุณาเลือกเมนูให้ถูกต้อง")
```

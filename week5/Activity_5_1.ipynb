**5. ตรวจสอบข้อมูลจำนวนมากอย่างรวดเร็ว**

Blocked IP Address

```python
blocked_ips = {
    "192.168.1.5",
    "192.168.1.8",
    "192.168.1.20"
}
```

ตรวจสอบ

```python
if "192.168.1.5" in blocked_ips:
    print("Blocked")
```

**เหตุผล**

- ตรวจสอบได้รวดเร็ว

- ไม่สนลำดับของข้อมูล

**ผู้เข้าชมเว็บไซต์ (Unique Visitors)**

```python
visitors = {
    "User001",
    "User002",
    "User003"
}
```

เมื่อมีคนเข้าเว็บไซต์

```python
visitors.add("User001")
visitors.add("User004")
```

ผลลัพธ์

```python
{'User001', 'User002', 'User003', 'User004'}
```

**เหตุผล**

- ไม่นับผู้เข้าชมซ้ำ

- เหมาะสำหรับเก็บข้อมูลที่ต้องไม่ซ้ำกั

```python
%%writefile my_utils.py

def greet(name):
    """แสดงข้อความทักทาย"""
    print(f"สวัสดี {name}!")


def is_prime(number):
    """ตรวจสอบว่าตัวเลขเป็นจำนวนเฉพาะหรือไม่"""

    if number < 2:
        return False

    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            return False

    return True
```

```python
import my_utils
import math
import random

my_utils.greet("Alice")

print("7 เป็นจำนวนเฉพาะหรือไม่:", my_utils.is_prime(7))
print("10 เป็นจำนวนเฉพาะหรือไม่:", my_utils.is_prime(10))

print("รากที่สองของ 64 คือ", math.sqrt(64))

print("ตัวเลขสุ่มระหว่าง 1 ถึง 100 คือ", random.randint(1, 100))
```

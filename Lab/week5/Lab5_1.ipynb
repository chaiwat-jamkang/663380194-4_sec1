# Lab 5.1: Simple Contact Book Manager (Dictionary Focus)

```python
# Lab 5.1 : Simple Contact Book Manager

contacts = {}

while True:
    print("\n===== Contact Book =====")
    print("1. Add/Update Contact")
    print("2. View Contact Details")
    print("3. List All Contacts")
    print("4. Delete Contact")
    print("5. Exit")

    choice = input("Choose an option (1-5): ")

    # 1. Add/Update Contact
    if choice == "1":
        name = input("Enter contact name: ").strip().lower()
        phone = input("Enter phone number: ")
        email = input("Enter email: ")

        if name in contacts:
            print("Contact exists. Updating information...")
        else:
            print("New contact added.")

        contacts[name] = {
            "phone": phone,
            "email": email
        }

    # 2. View Contact Details
    elif choice == "2":
        name = input("Enter contact name: ").strip().lower()

        contact = contacts.get(name)

        if contact:
            print("\nContact Details")
            print("Name :", name.title())
            print("Phone:", contact["phone"])
            print("Email:", contact["email"])
        else:
            print("Contact not found.")

    # 3. List All Contacts
    elif choice == "3":
        if len(contacts) == 0:
            print("No contacts available.")
        else:
            print("\n===== Contact List =====")
            for name, info in contacts.items():
                print(f"Name : {name.title()}")
                print(f"Phone: {info['phone']}")
                print(f"Email: {info['email']}")
                print("-" * 25)

    # 4. Delete Contact
    elif choice == "4":
        name = input("Enter contact name: ").strip().lower()

        removed = contacts.pop(name, None)

        if removed:
            print("Contact deleted successfully.")
        else:
            print("Contact not found.")

    # 5. Exit
    elif choice == "5":
        print("Exiting program...")
        break

    else:
        print("Invalid choice. Please select 1-5.")
```

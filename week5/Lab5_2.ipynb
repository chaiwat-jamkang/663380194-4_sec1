```python
# Lab 5.2 : Course Enrollment & Collaboration

# Initialize Sets
python_students = {"S001", "S003", "S005", "S007", "S009"}
java_students = {"S002", "S003", "S006", "S007", "S010"}

# 1. Students in Both Courses (Intersection)
both_courses = python_students.intersection(java_students)
print("Students in both Python and Java:")
print(both_courses)

# 2. All Unique Students (Union)
all_students = python_students.union(java_students)
print("\nAll unique students:")
print(all_students)

# 3. Students Only in Python (Difference)
python_only = python_students.difference(java_students)
print("\nStudents only in Python:")
print(python_only)

# 4. Students in Only One Course (Symmetric Difference)
one_course_only = python_students.symmetric_difference(java_students)
print("\nStudents enrolled in only one course:")
print(one_course_only)

# 5. Remove Duplicates from a List
all_students_list = ["S001", "S003", "S005", "S001", "S002"]

unique_students_list = list(set(all_students_list))

print("\nOriginal List:")
print(all_students_list)

print("\nUnique List:")
print(unique_students_list)
```

```python

import requests
from bs4 import BeautifulSoup

url = "https://automatetheboringstuff.com/2e/"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers, timeout=10)
response.raise_for_status()

print("Successfully downloaded content.")

soup = BeautifulSoup(response.text, "html.parser")

# ดึงชื่อหนังสือ โดยลองค้นหาจาก h1 ก่อน
book_title_tag = soup.find("h1")

if book_title_tag:
    book_title = book_title_tag.get_text(strip=True)
else:
    book_title = soup.title.get_text(strip=True) if soup.title else "Book Title Not Found"

print("\n--- Scraped Data ---")
print("Book Title:", book_title)

# ดึงชื่อ Chapter
print("\nChapter Titles:")

chapter_titles = []
toctree_wrapper = soup.find("div", class_="toctree-wrapper")

if toctree_wrapper:
    chapter_links = toctree_wrapper.select(
        "li.toctree-l1 a.reference.internal"
    )

    for link in chapter_links:
        title = link.get_text(strip=True)
        if title:
            chapter_titles.append(title)

# ถ้าไม่พบโครงสร้าง toctree ให้ค้นหาลิงก์ Chapter จากทั้งหน้า
if not chapter_titles:
    for link in soup.find_all("a", href=True):
        title = link.get_text(" ", strip=True)

        if title.lower().startswith("chapter ") and title not in chapter_titles:
            chapter_titles.append(title)

if chapter_titles:
    for i, title in enumerate(chapter_titles, start=1):
        print(f"{i}. {title}")
else:
    print("No chapter titles found.")

print("--------------------")
```

```python

# Activity 6.1: Refactor Lab 3.2 Using Functions

def countdown(start):
    """นับถอยหลังจากตัวเลขที่กำหนดจนถึง 0"""

    while start >= 0:
        print(start)
        start -= 1

    print("Blast off!")


def check_guess(secret_number, guess):
    """ตรวจสอบผลการทายตัวเลข"""

    if guess < secret_number:
        return f"{guess} ต่ำเกินไป! ลองอีกครั้ง"
    elif guess > secret_number:
        return f"{guess} สูงเกินไป! ลองอีกครั้ง"
    else:
        return "ยินดีด้วย! คุณทายถูกแล้ว!"


def guessing_game():
    """เริ่มเกมทายตัวเลข"""

    secret_number = 42

    while True:
        guess = int(input("ทายตัวเลข: "))

        result = check_guess(secret_number, guess)
        print(result)

        if guess == secret_number:
            break


# ===== Main Program =====
start = int(input("กรุณากรอกตัวเลขเริ่มต้นสำหรับนับถอยหลัง: "))
countdown(start)

guessing_game()
```

การปรับปรุงโค้ด (Refactoring)

- สร้างฟังก์ชัน countdown() สำหรับการนับถอยหลัง

- สร้างฟังก์ชัน check_guess() สำหรับตรวจสอบผลการทายตัวเลข

- สร้างฟังก์ชัน guessing_game() สำหรับควบคุมการทำงานของเกม

- เพิ่ม Docstring เพื่ออธิบายหน้าที่ของแต่ละฟังก์ชัน
